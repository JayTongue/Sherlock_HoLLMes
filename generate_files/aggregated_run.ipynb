```python
from pathlib import Path
import json
from tqdm import tqdm

from gen_cont_complex import gen_cont_complex_main
```

```python
folder_tiers = {
    10: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2}, 
    25: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2},
    50: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2}, 
    100: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2}, 
    250: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2}, 
    500: {'simple_retrieval': 2, 'formal_logic': 2, 'informal_logic': 2}
                }
num_trials = 10
problem_types = ['zeros', 'random', 'markov', 'enron', 'contracts']

root = 'data/sized'
```

```python
for folder_tier in tqdm(folder_tiers):
    for i in tqdm(range(num_trials), desc=folder_tier):
        aggregated_problems = {}
        for problem_type in tqdm(problem_types):
            final_path = Path(f'{root}/{folder_tier}/{i}/{problem_type}')
            # print(f'Working on: {final_path}')
            final_path.mkdir(parents=True, exist_ok=True)
            problem = gen_cont_complex_main(problem_type, final_path, folder_tier, folder_tiers[folder_tier])
            aggregated_problems[problem_type] = problem
        with open(f'{root}/{folder_tier}/{i}/key_{folder_tier}_{i}.json', 'w') as outfile:
            json.dump(aggregated_problems, outfile, indent=4)
```
